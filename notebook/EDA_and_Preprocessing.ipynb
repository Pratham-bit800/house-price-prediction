# 4. Exploratory Data Analysis

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.histplot(data['price'], bins=50, kde=True, ax=axes[0])
axes[0].set_title('Distribution of House Prices', fontsize=13)
axes[0].set_xlabel('Price (USD)')
axes[0].set_ylabel('Frequency')

sns.boxplot(x=data['price'], ax=axes[1])
axes[1].set_title('Boxplot of House Prices', fontsize=13)
axes[1].set_xlabel('Price (USD)')

plt.tight_layout()
plt.show()
```

```python
# Heatmap

# Compute correlation matrix for numerical columns
numeric_df = data.select_dtypes(include=[np.number])
corr_matrix = numeric_df.corr()

plt.figure(figsize=(14, 10))
sns.heatmap(corr_matrix,
            annot=True,
            fmt='.2f',
            cmap='coolwarm',
            linewidths=0.5)
plt.title('Correlation Heatmap of Numerical Features')
plt.tight_layout()
plt.show()
```

```python
# Histogram 
num_features = ['bedrooms', 'bathrooms', 'sqft_living', 'sqft_lot', 'floors', 'grade']

fig, axes = plt.subplots(2, 3, figsize=(16, 9))
axes = axes.flatten()

for i, feature in enumerate(num_features):
    sns.histplot(data[feature], bins=30, kde=True, ax=axes[i])
    axes[i].set_title(f'Distribution of {feature}', fontsize=11)
    axes[i].set_xlabel(feature)
    axes[i].set_ylabel('Frequency')

plt.tight_layout()
plt.show()
```

```python
# Scatterplot

fig, axes = plt.subplots(2, 2, figsize=(14, 10))

sns.scatterplot(x=data['sqft_living'], y=data['price'], ax=axes[0, 0], alpha=0.4)
axes[0, 0].set_title('Living Area vs Price')
axes[0, 0].set_xlabel('sqft_living')
axes[0, 0].set_ylabel('price')

sns.scatterplot(x=data['grade'], y=data['price'], ax=axes[0, 1], alpha=0.4)
axes[0, 1].set_title('Grade vs Price')
axes[0, 1].set_xlabel('grade')
axes[0, 1].set_ylabel('price')

sns.scatterplot(x=data['yr_built'], y=data['price'], ax=axes[1, 0], alpha=0.4)
axes[1, 0].set_title('Year Built vs Price')
axes[1, 0].set_xlabel('yr_built')
axes[1, 0].set_ylabel('price')

sns.scatterplot(x=data['lat'], y=data['price'], ax=axes[1, 1], alpha=0.4)
axes[1, 1].set_title('Latitude vs Price')
axes[1, 1].set_xlabel('lat')
axes[1, 1].set_ylabel('price')

plt.tight_layout()
plt.show()
```

# 5. Feature Engineering

```python
# Extract sale year and sale month from the date column
data['sale_year'] = data['date'].dt.year
data['sale_month'] = data['date'].dt.month

# House age at the time of sale
data['house_age'] = data['sale_year'] - data['yr_built']

# Renovation related features
data['was_renovated'] = (data['yr_renovated'] > 0).astype(int)
data['years_since_renovation'] = np.where(
    data['was_renovated'] == 1,
    data['sale_year'] - data['yr_renovated'],
    data['house_age']  # if never renovated, use house_age as a proxy
)

# Total room count
data['total_rooms'] = data['bedrooms'] + data['bathrooms']

# New columns
data[['sale_year', 'sale_month', 'house_age', 'was_renovated', 'years_since_renovation', 'total_rooms']].head()
```

```python
# Dropping unecessary columns
columns_to_drop = ['id', 'date', 'yr_built', 'yr_renovated', 'zipcode']
df_model = data.drop(columns=columns_to_drop)

print("Remaining columns:")
print(list(df_model.columns))
```

```python
# Export cleaned dataset for model training
df_model.to_csv("../data/cleaned_dataset.csv", index=False)
print(f"Cleaned dataset exported: {df_model.shape[0]:,} rows x {df_model.shape[1]} columns")
```
